```python
import numpy as np

arr1= np.array([1,2,3])
print("numpy_array1",arr1)

arr1.shape
```

```python
arr2= np.array([[1,2,3],[4,5,6]])
print("numpy_array2",arr2)

arr2.shape
```

```python
#dtype
a = np.array([3,6,2,7])
print(a.dtype)
```

```python
#reshape()
a = np.array([1,2,3,4,5,6,])
b=a.reshape(2,3)
print(b)
```

```python
#astype()
a = np.array([1,2,3,4])
b = a.astype(float)
print("float",b)
```

```python
#zeros() , ones()
z = np.zeros(2)
o = np.ones((2,3))
print("vector float zeros",z)
print("matrix float ones",o)
```

```python
#ndim
a=np.array([1,2,3])
b=np.array([[1,2,3],[4,5,6]])

print(a.ndim,"D array")
print(b.ndim,"D array")
```

```python
#arange()
a=np.arange(1,6) # (1= include, 6= exclude)
print("arange",a)
```

```python
#sum() ,mean(), max(), min()
a = np.array([1,2,3,4])
print(a.sum())
print(a.mean())
print(a.max())
print(a.min())
```

```python
#flatten()
a = np.array([[1,2,3,4],[5,6,3,6]])
print("flatten_arr",a.flatten())

```

```python
# T (Transpose)
a = np.array([[1,2],[5,6]])
print("T_arr",a.T)

```

```python
#dot() @
a = np.array([[1,2],[5,6]])
b = np.array([[2,3],[4,2]])
print(a.dot(b))

```

```python
#random.rand() 1.0 >= 0.0
a= np.random.rand(2,3)
print("random_rand",a)
```
